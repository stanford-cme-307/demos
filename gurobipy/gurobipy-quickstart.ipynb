```python
# import everything from gurobi
from gurobipy import *
```

```python
# Generate problem data 
import numpy as np
np.random.seed(1)
m = 5
n = 10
A = np.random.randn(m,n)
b = np.random.randn(m)
c = np.random.randn(n)
l = np.zeros(n)
u = np.ones(n)

# check that A has full row rank
assert np.linalg.matrix_rank(A) == m
```

```python
# solve the problem min c^T x
# subject to Ax = b, x >= 0 

model = Model()
x = model.addMVar(shape=n, lb=0.0, name="x")
model.setObjective(c @ x, GRB.MINIMIZE)
model.addConstr(A @ x == b)
model.optimize()
```

```python
# solve the problem min c^T x
# subject to Ax <= b
#            l <= x <= u

model = Model()
x = model.addMVar(shape=n, name="x")
model.setObjective(c @ x, GRB.MINIMIZE)
model.addConstr(A @ x <= b)
model.addConstr(x >= l)
model.addConstr(x <= u)
model.optimize()
```
